```python
import pdfplumber
import pandas as pd
import os

PARAMETERS = [
    "Power KW",
    "Frequency KHZ",
    "Voltage V",
    "Current A",
    "Energy KWS",
    "Temperature *c",
    "Spare1",
    "Spare1_2",
    "Quench Temperature",
    "Quench Flow1",
    "Quench Flow2",
    "Quench Flow3",
    "Quench Flow4",
    "Quench Flow5",
    "Quench Flow6",
    "Quench Flow7",
    "Quench Flow8",
    "Quench Flow9",
    "Heat Cycle",
    "Quench Cyle",
    "Cycle TIme",
    "Heat Error",
    "Quench Error",
    "Result",
]

def parse_parameter_row(line):
    KNOWN_PARAMS = [
        "Quench Temperature",
        "Quench Flow1",
        "Heat Cycle",
        "Quench Cyle",
        "Cycle TIme",
        "Heat Error",
        "Quench Error",
        "Power KW",
        "Frequency KHZ",
        "Voltage V",
        "Current A",
        "Energy KWS",
        "Temperature *c",
        "Spare1",
        "Result",
    ]

    for param in KNOWN_PARAMS:
        if line.startswith(param):
            remainder = line[len(param):].strip().split()
            if len(remainder) >= 2:
                return (param, remainder[1])
            elif len(remainder) == 1:
                return (param, remainder[0])
    return None


def extract_pdf_data(filename, filepath):
    """
    ✅ Fixed version — opens PDF directly from file path (string)
    """
    record = {"File Name": filename}

    batch_fields = {
        "Part Name":     None,
        "Serial No":     None,
        "Result":        None,
        "Date & Time":   None,
        "Referance No1": None,
        "Operator":      None,
    }

    try:
        # ✅ KEY FIX — Open directly from path, no BytesIO needed
        with pdfplumber.open(filepath) as pdf:
            pages_to_read = pdf.pages[:2]

            for page_index, page in enumerate(pages_to_read):
                text  = page.extract_text()
                lines = text.split("\n") if text else []

                section     = "HARDENING" if page_index == 0 else "TEMPERING"
                spare_count = 0
                in_table    = False

                for line in lines:
                    line = line.strip()
                    if not line:
                        continue

                    # ── Batch Details (Page 1 only) ──
                    if page_index == 0:
                        if line.startswith("Part Name"):
                            parts = line.split()
                            try:
                                batch_fields["Part Name"] = parts[2]
                                sn_idx = parts.index("No", 3)
                                batch_fields["Serial No"] = parts[sn_idx + 1]
                            except (IndexError, ValueError):
                                pass

                        elif line.startswith("Result"):
                            parts = line.split()
                            try:
                                batch_fields["Result"]      = parts[1]
                                batch_fields["Date & Time"] = parts[5] + " " + parts[6]
                            except IndexError:
                                pass

                        elif line.startswith("Referance No1"):
                            parts = line.split()
                            try:
                                batch_fields["Referance No1"] = parts[1]
                            except IndexError:
                                pass

                        elif line.startswith("Customer"):
                            parts = line.split()
                            try:
                                op_idx = parts.index("Operator")
                                if op_idx + 1 < len(parts):
                                    batch_fields["Operator"] = " ".join(parts[op_idx + 1:])
                            except ValueError:
                                pass

                    # ── Table Header ──
                    if line.startswith("Parameter"):
                        in_table    = True
                        spare_count = 0
                        continue

                    # ── Skip page markers ──
                    if line.startswith("--") and line.endswith("--"):
                        continue

                    # ── Parse Rows ──
                    if in_table:
                        parsed = parse_parameter_row(line)
                        if parsed:
                            param_name, actual_val = parsed
                            if param_name == "Spare1":
                                spare_count += 1
                                if spare_count == 2:
                                    param_name = "Spare1_2"
                            col_name = f"{section}_{param_name}"
                            record[col_name] = actual_val

    except Exception as e:
        print(f"  ⚠️  Error reading {filename}: {e}")

    record.update(batch_fields)
    return record

print("✅ Functions loaded successfully!")
```

```python
# ── ZIP version of Cell 4 ──
print(f"🔄 Processing {len(pdf_files_list)} PDF file(s)...\n")
all_records = []

for i, filename in enumerate(pdf_files_list, 1):
    filepath = os.path.join(extract_folder, filename)   # 👈 Key fix — uses path not bytes
    print(f"  📄 [{i}/{len(pdf_files_list)}] {filename}")
    record = extract_pdf_data(filename, filepath)
    all_records.append(record)

print(f"\n✅ All files processed!")
```

```python
# ── Build DataFrame ──
df = pd.DataFrame(all_records)

# ── Define column order ──
info_cols = ["File Name", "Part Name", "Serial No",
             "Date & Time", "Referance No1", "Operator", "Result"]

hardening_cols = [f"HARDENING_{p}" for p in PARAMETERS
                  if f"HARDENING_{p}" in df.columns]
tempering_cols = [f"TEMPERING_{p}" for p in PARAMETERS
                  if f"TEMPERING_{p}" in df.columns]

info_cols      = [c for c in info_cols      if c in df.columns]
hardening_cols = [c for c in hardening_cols if c in df.columns]
tempering_cols = [c for c in tempering_cols if c in df.columns]

final_cols = info_cols + hardening_cols + tempering_cols
df = df[final_cols]

# ── Rename columns ──
rename_map = {}
for col in hardening_cols:
    rename_map[col] = col.replace("HARDENING_", "H_")
for col in tempering_cols:
    rename_map[col] = col.replace("TEMPERING_", "T_")
df.rename(columns=rename_map, inplace=True)

# ── Preview in Colab ──
print("📊 Preview of extracted data:\n")
display(df)

# ── Save to Excel ──
output_filename = "GH_Actuals_Report.xlsx"

with pd.ExcelWriter(output_filename, engine="openpyxl") as writer:

    # Sheet 1: All Data
    df.to_excel(writer, sheet_name="All Data", index=False)

    # Sheet 2: Hardening Only
    h_renamed = [rename_map.get(c, c) for c in hardening_cols]
    h_cols    = [c for c in info_cols] + [c for c in h_renamed if c in df.columns]
    df[h_cols].to_excel(writer, sheet_name="Hardening", index=False)

    # Sheet 3: Tempering Only
    t_renamed = [rename_map.get(c, c) for c in tempering_cols]
    t_cols    = [c for c in info_cols] + [c for c in t_renamed if c in df.columns]
    df[t_cols].to_excel(writer, sheet_name="Tempering", index=False)

    # ── Auto-fit column widths ──
    for sheet_name in writer.sheets:
        ws = writer.sheets[sheet_name]
        for col_cells in ws.columns:
            max_len = max(
                (len(str(cell.value)) for cell in col_cells if cell.value),
                default=10
            )
            ws.column_dimensions[col_cells[0].column_letter].width = max_len + 4

# ── Auto Download Excel ──
files.download(output_filename)

print(f"\n✅ Excel file '{output_filename}' downloaded successfully!")
print(f"   📊 Total Records    : {len(df)}")
print(f"   📊 Hardening Cols   : {len(h_renamed)}")
print(f"   📊 Tempering Cols   : {len(t_renamed)}")
```
